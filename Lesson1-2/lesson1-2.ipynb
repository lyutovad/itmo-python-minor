### Имена переменных

В учебном коде удобно придерживаться следующих правил:

- использовать содержательные имена: `student_age`, а не `x1`;
- писать составные имена в стиле `snake_case`;
- не начинать имя с цифры;
- не использовать пробелы и дефисы;
- не переопределять без необходимости встроенные функции, например `list`, `str`, `sum`, `max`.

У каждой переменной в Python должно быть уникальное имя. Можно использовать буквы a–z, цифры 0–9, символ _.

Стиль snake_case - разделение слов знаком подчеркивания (my_variable).

Первым символом не может быть цифра, т.к. Python — язык регистрозависимый (большие и маленькие буквы в имени переменной различаются).

Неправильно: 5_var, User_1, 0data, tmpFile, minStep_shift

Отдельного типа для констант в Python нет, но, как и во многих других языках, есть соглашение между
разработчиками — писать константы в верхнем регистре, например: MAX_ATTEMPTS, MEDIA_URL, DEBUG

* давайте переменным осмысленные имена
* коротких переменных вида a, b, x, y, z - в особых случаях
* никакого транслита : skorost, imya. Только английские слова.

**!!! Никода не использовать в качестве имен:**

[Зарезервированные слова](https://docs.python.org/3/reference/lexical_analysis.html#keywords)

[Встроенные функции](https://docs.python.org/3/library/functions.html)

[Встроенные типы](https://docs.python.org/3/library/stdtypes.html)

### Изменяемость

Список можно изменить после создания. Это важно помнить, когда одна и та же структура используется в нескольких местах программы.

````python
numbers = [1, 2, 3]
other_numbers = numbers
other_numbers.append(4)

print(numbers)
````

Чтобы создать отдельную поверхностную копию списка, можно использовать `.copy()`.

````python
numbers = [1, 2, 3]
other_numbers = numbers.copy()
other_numbers.append(4)

print(numbers)
print(other_numbers)
````

## 6. Кортежи и множества

### Кортеж

Кортеж похож на список, но его элементы нельзя заменить, добавить или удалить после создания.

````python
point = (59.94, 30.31)
print(point[0])
print(point[1])
````

Кортежи удобны для небольших наборов связанных значений, которые не предполагается изменять.

### Множество

Множество хранит только уникальные элементы и не предназначено для обращения по позиции.

````python
cities = ["Москва", "Казань", "Москва", "Томск", "Казань"]
unique_cities = set(cities)
print(unique_cities)
````

### Операции над множествами

````python
group_a = {"Анна", "Иван", "Ольга"}
group_b = {"Иван", "Ольга", "Пётр"}

print(group_a & group_b)  # пересечение
print(group_a | group_b)  # объединение
print(group_a - group_b)  # разность
````

### Значение параметра по умолчанию

````python
def calculate_total(price, count, discount=0):
    total = price * count
    return total * (1 - discount)

print(calculate_total(100, 3))
print(calculate_total(100, 3, discount=0.10))
````

### `return` и `print` — не одно и то же

`print` выводит значение на экран. `return` возвращает результат функции, чтобы его можно было использовать дальше.

````python
def square_with_print(number):
    print(number ** 2)

def square_with_return(number):
    return number ** 2

result_1 = square_with_print(5)
result_2 = square_with_return(5)

print("result_1:", result_1)
print("result_2:", result_2)
````

## Область видимости переменных

Переменные, созданные внутри функции, обычно существуют только внутри этой функции.

Также параметры функции являются локальными переменными.

````python
def calculate_mean(values):
    total = sum(values)
    mean_value = total / len(values)

    print("Внутри функции:", total, mean_value)

    return mean_value


result = calculate_mean([10, 20, 30])

print("Результат:", result)
````

Внутри функции доступны:

- `values` — параметр функции;
- `total` — переменная, созданная внутри функции;
- `mean_value` — переменная, созданная внутри функции.

После завершения функции мы получаем только то значение, которое было передано через `return`.

Переменные `total` и `mean_value` снаружи функции недоступны.

````python
# print(total)
````

Если раскомментировать строку выше, возникнет `NameError`, потому что `total` существует только внутри функции.

### Обработка ожидаемой ошибки

Если ошибка является ожидаемой частью данных, её можно обработать конструкцией `try` — `except`.

````python
def parse_age(value):
    try:
        return int(value)
    except ValueError:
        return None

print(parse_age("25"))
print(parse_age("неизвестно"))
````

## 18. Короткие функции `lambda`

Иногда простую функцию используют один раз, например как правило сортировки. Для этого существует конструкция `lambda`.

Обычная функция:

````python
def get_age(student):
    return student["age"]
````

Та же логика как короткая функция:

````python
get_age_short = lambda student: student["age"]
print(get_age_short({"name": "Анна", "age": 25}))
````

Типичный прикладной пример — сортировка:

````python
students = [
    {"name": "Анна", "age": 25},
    {"name": "Иван", "age": 31},
    {"name": "Ольга", "age": 22},
]

sorted_students = sorted(students, key=lambda student: student["age"])
print(sorted_students)
````

`lambda` не заменяет обычные функции. Если логика длиннее одного простого выражения или требует пояснения, лучше использовать `def`.

Массив — упорядоченная изменяемая структура данных, к элементам которой можно обратиться по индексу

Основной объект NumPy является **ndarray** (n-dimensional array) - многомерный однородный массив с заранее заданным количеством элементов. 
* все объекты в нем одного размера или типа (тип данных определен другим объектом NumPy, который называется dtype (тип-данных)). Каждый ndarray ассоциирован только с одним типом dtype.
* размер массива фиксирован, а это значит, что после создания объекта его уже нельзя поменять. Это поведение отличается от такового у списков Python, которые могут увеличиваться и уменьшаться в размерах.


Основные атрибуты объекта numpy.ndarray в Python:

1. .shape: Это кортеж, показывающий размеры массива по каждой оси. В данном случае (3, 5), что означает 3 строки и 5 столбцов.

2. .ndim: Количество измерений или осей массива. Здесь 2, так как массив двумерный.

3. .dtype: Тип данных элементов массива. В этом примере int32, что означает 32-битное целое число.

4. .itemsize: Размер каждого элемента в байтах. Здесь 4 байта, что соответствует типу int32.

5. .size: Общее количество элементов в массиве. В данном случае 15 (3 * 5).



## 1. Зачем NumPy, если уже есть списки?

Списки Python — универсальная структура. В них можно хранить почти любые объекты.

NumPy создаёт массивы, ориентированные на численные данные и одинаковые операции над большим количеством значений.

Сравним поведение списка и массива.


````python
values_list = [10, 20, 30]

print(values_list * 2)
````

Для списка оператор `*` означает повторение списка.

Для массива NumPy тот же оператор применяется **к каждому элементу**.


````python
values = np.array([10, 20, 30])

print(values * 2)
````

Это одна из главных идей NumPy: многие операции выполняются сразу над всем массивом.

Такую работу часто называют **векторизованной операцией**.

## 2. `ndarray`: тип, форма и размер

Основной объект NumPy — `numpy.ndarray`.


````python
values = np.array([10, 20, 30, 40])

print(type(values))
print(values.dtype)
print(values.shape)
print(values.ndim)
print(values.size)
````

## 6. Сравнения и логические маски

Сравнение массива с числом тоже происходит поэлементно.


````python
scores = np.array([45, 67, 82, 51, 90, 73])

mask = scores >= 60

print(mask)
````

Получился массив значений `True` / `False`.

Такой массив можно использовать как **маску** для выбора элементов.


````python
passed_scores = scores[scores >= 60]

print(passed_scores)
````

Это очень важный переход к Pandas. Позже мы будем писать почти ту же конструкцию для таблиц:

```python
data[data["score"] >= 60]
```


### Несколько условий

Для массивов NumPy условия объединяют операторами `&` и `|`.

Каждое условие лучше заключать в круглые скобки.


````python
scores = np.array([45, 67, 82, 51, 90, 73])

selected = scores[(scores >= 60) & (scores < 80)]
print(selected)
````

## 7. Двумерный массив

Массив может иметь больше одного измерения.


````python
data = np.array([
    [25, 1200],
    [31, 2300],
    [22, 900],
])

print(data)
print(data.shape)
print(data.ndim)
````

Здесь `shape == (3, 2)`:

- 3 строки;
- 2 столбца.

Но NumPy не знает, что первый столбец — возраст, а второй — сумма покупки. Для него это просто числа.


````python
print(data[0])      # первая строка
print(data[0, 1])   # первая строка, второй столбец
print(data[:, 0])   # все строки, первый столбец
````

## 8. Почему NumPy недостаточно для обычной таблицы?

Представим данные о клиентах:

```text
имя | возраст | город | покупка
```

У столбцов разные смыслы и даже разные типы данных:

- имя — строка;
- возраст — число;
- город — строка;
- покупка — число.

Для такой структуры удобнее использовать **Pandas DataFrame**.


## 9. Создаём DataFrame

Вернёмся к уже знакомому списку словарей.


````python
records = [
    {"name": "Анна", "age": 25, "city": "Москва", "purchase": 1200.5},
    {"name": "Иван", "age": 17, "city": "Казань", "purchase": 700.0},
    {"name": "Ольга", "age": 31, "city": "Москва", "purchase": 1800.0},
    {"name": "Пётр", "age": 44, "city": "Томск", "purchase": 2300.0},
]
````

````python
import pandas as pd

data = pd.DataFrame(records)

data
````

Разберём выражение:

```python
data = pd.DataFrame(records)
```

- `pd` — псевдоним модуля `pandas`;
- `DataFrame` — класс;
- `pd.DataFrame(...)` — вызов класса;
- результат — объект `DataFrame`;
- объект сохраняется в переменной `data`.

То есть объектная модель из базового Python здесь начинает работать практически.

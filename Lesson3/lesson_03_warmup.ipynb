## 10. Небольшая связная задача

Что делает этот код?

```python
users = [
    {"name": "Анна", "city": " Москва "},
    {"name": "Иван", "city": "Казань"},
    {"name": "Ольга", "city": "МОСКВА"},
]

moscow_users = []

for user in users:
    city = user["city"].strip().lower()

    if city == "москва":
        moscow_users.append(user["name"])

print(moscow_users)
```

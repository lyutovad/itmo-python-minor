# Часть 1. Заканчиваем NumPy

## 1. Несколько условий

В обычном Python мы писали `and` / `or`. В NumPy сравнение создаёт сразу массив `True` / `False`, поэтому используются поэлементные операторы:

- `&` — и;
- `|` — или.

Каждое условие берём в круглые скобки.

````python
import numpy as np

ages = np.array([17, 18, 25, 31, 16, 40])

selected = ages[(ages >= 18) & (ages <= 30)]
print(selected)
````

### Сколько элементов подходит?

Булевы значения можно суммировать: `True` считается как `1`, `False` — как `0`.

````python
adult_mask = ages >= 18

print(adult_mask)
print(adult_mask.sum())
````

## 2. Двумерный массив

Двумерный массив можно представить как прямоугольную таблицу чисел.

````python
measurements = np.array([
    [18, 60],
    [21, 55],
    [19, 70],
    [25, 65],
])

print(measurements)
print(measurements.shape)
print(measurements.ndim)
````

Если `shape == (4, 2)`, это 4 строки и 2 столбца.

````python
print(measurements[0])      # первая строка
print(measurements[0, 1])   # строка 0, столбец 1
print(measurements[:, 0])   # все строки, столбец 0
print(measurements[:, 1])   # все строки, столбец 1
````

NumPy видит строки и столбцы, но не знает, что означает каждый столбец. Для обычных таблиц с именами столбцов удобнее Pandas.

## 10. Несколько условий

Используем `&` и `|`, каждое условие берём в скобки.

````python
moscow_paid = orders[
    (orders["city"] == "Москва")
    & (orders["status"] == "paid")
]

moscow_paid
````

````python
selected = orders[
    (orders["amount"] >= 1500)
    & (orders["amount"] <= 3000)
]

selected
````

Почему не `and`? Потому что сравнение столбца создаёт целую `Series` из логических значений. `&` и `|` работают поэлементно.

## 13. `sort_values()`

## 14. Почему после фильтрации индекс может быть «с дырками»?

````python
expensive_orders = orders[orders["amount"] > 2500]

expensive_orders
````

Pandas сохраняет исходные индексы. Если нужен новый индекс `0, 1, 2, ...`:

````python
expensive_orders = expensive_orders.reset_index(drop=True)

expensive_orders
````

`drop=True` означает: старый индекс не превращаем в отдельный столбец.
